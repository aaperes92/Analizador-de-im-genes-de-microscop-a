Pruebas de procesamiento de imagenes SEM con skimage

```python
import numpy
import skimage as ski
import matplotlib.pyplot as plt
from pathlib import Path
import exifread 
```

### funciones

```python
# funcion de carga de imagenes

def cargar_imagen(path: Path) -> np.ndarray:
    """Carga imagen SEM como array 2D de floats."""
    return ski.skio.imread(str(path), as_gray=True)


def extraer_exif(path: Path) -> dict:
    """Devuelve diccionario con los tags EXIF de la imagen."""
    with open(path, 'rb') as f:
        tags = exifread.process_file(f)
    return {str(k): str(v) for k, v in tags.items()}
```
